```python
import matplotlib.pyplot as plt

daftarMahasiswa = []
jumlahMahasiswa = int(input("Berapa Jumlah Mahasiswa Kelas A = "))

for i in range(jumlahMahasiswa):
    namaMahasiswa = input(f"\nMasukkan Nama Mahasiswa ke-{i+1}: ")
    daftarMahasiswa.append(namaMahasiswa)

    # Input Nilai
    daftarNilai_Mahasiswa = []
    jumlahNilaiMahasiswa = int(input(f"Jumlah Penilaian {namaMahasiswa} dalam 1 semester = "))

    for j in range(jumlahNilaiMahasiswa):
        nilaiMahasiswa = float(input(f"  > Nilai ke-{j+1} : "))
        daftarNilai_Mahasiswa.append(nilaiMahasiswa)

    # Kalkulasi
    nilaiTertinggi = max(daftarNilai_Mahasiswa)
    nilaiTerendah = min(daftarNilai_Mahasiswa)
    nilaiRata2 = sum(daftarNilai_Mahasiswa) / len(daftarNilai_Mahasiswa)

    # Penentuan Status
    if nilaiRata2 > 70:
        status = "LULUS"
        warna_status = 'green'
    else:
        status = "TIDAK LULUS"
        warna_status = 'red'

    print(f"\n--- Ringkasan Nilai {namaMahasiswa} ---")
    print(f"Tertinggi: {nilaiTertinggi} | Terendah: {nilaiTerendah} | Rata-rata: {nilaiRata2:.2f}")
    print(f"Status: {status}")

    # --- PEMBUATAN GRAFIK ---
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    fig.suptitle(f"Analisis Nilai: {namaMahasiswa}", fontsize=16)

    # 1. Grafik Batang Nilai Tertinggi vs Terendah
    ax1.bar(['Tertinggi', 'Terendah'], [nilaiTertinggi, nilaiTerendah], color=['teal', 'coral'])
    ax1.set_ylim(0, 100)
    ax1.set_title("Perbandingan Nilai Ekstrem")
    ax1.set_ylabel("Skala Nilai")

    # 2. Grafik Pie Data Kelulusan (Berdasarkan list nilai mahasiswa tersebut)
    lulus = len([n for n in daftarNilai_Mahasiswa if n >= 70])
    gagal = len(daftarNilai_Mahasiswa) - lulus

    ax2.pie([lulus, gagal], labels=['Lulus (>=70)', 'Gagal (<70)'],
            autopct='%1.1f%%', colors=['#66b3ff','#ff9999'], startangle=140)
    ax2.set_title("Proporsi Kelulusan per Item Nilai")

    plt.tight_layout()
    plt.show()
```
